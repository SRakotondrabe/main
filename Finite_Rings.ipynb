```python
from itertools import product
import numpy as np
import random
 
```

```python
# GENERATE MULTIPLICATION TABLE


def tuple_to_int(t, modul): # encode in {1,2,...,N}
    index = 1  
    multiplier = 1 
    for i, val in enumerate(t):
        index += val * multiplier
        multiplier *= modul[i]
    return index


def multiplication_table(modul):

    elements = list(product(*[range(n) for n in modul]))
    size = len(elements)

    encoding_map = {elements[i]: tuple_to_int(elements[i], modul) for i in range(size)}

    table = [[None] * size for _ in range(size)]

    for i in range(size):
        for j in range(size):
            elt = tuple((elements[i][d] * elements[j][d]) % modul[d] for d in range(len(modul)))
            table[i][j] = encoding_map[elt]

    return np.array(table)
```

```python
# GENERATE ADDITION TABLE
def addition_table(modul):

    elements = list(product(*[range(n) for n in modul]))
    size = len(elements)

    encoding_map = {elements[i]: tuple_to_int(elements[i], modul) for i in range(size)}

    table = [[None] * size for _ in range(size)]

    for i in range(size):
        for j in range(size):
            elt = tuple((elements[i][d] + elements[j][d]) % modul[d] for d in range(len(modul)))
            table[i][j] = encoding_map[elt]

    return np.array(table)
```

```python
# ROW PERMUTATION
def row_permuitation(mat1, mat2):
    mat1 = list(mat1)
    mat2 = list(mat2)
    n = len(mat1)
    idx = random.sample(list(range(n)), n)
    mat1 = [mat1[i] for i in idx]
    mat2 = [mat2[i] for i in idx]
    return mat1, mat2

```

```python
# COLONE PERMUTATION
def col_permutation(mat1, mat2):

    L1 = list(zip(*mat1))
    L2 = list(zip(*mat2))
    n = len(L1)
    index = random.sample(list(range(n)), n)
    L1 = [L1[i] for i in index]
    L2 = [L2[i] for i in index]

    mat1 = list(zip(*L1))
    mat2 = list(zip(*L2)) 
    return mat1, mat2




    
```

```python
def fill(mat, n):
    M = np.zeros((n,n))
    a, b = mat.shape
    M[:a, :b] = mat  
    return M
```

```python
def prime_factor(n):
    factor = []
    while n%2 == 0:
        factor.append(2)
        n = n//2
    d = 3
    while d <= n:
        if n%d == 0:
            factor.append(d)
            n = n//d
        else:
            d = d+2
    return factor
```

```python
# GENERATE RANDOM ROW\COL PERMUTATION
# ex: G = [2, 2, 3] ----> (Z/2Z)*(Z/2Z)*(Z/3Z)

def k_random(k, G):
    table_mult = fill(multiplication_table(G),N)
    table_add = fill(addition_table(G),N)
    data = [[table_mult, table_add]]
    for _ in range(k):
        Mpr, Apr = row_permuitation(table_mult, table_add)
        data.append([np.array(Mpr), np.array(Apr)])
    for _ in range(k):
        Mpc, Apc = row_permuitation(table_mult ,table_add)
        data.append([np.array(Mpc), np.array(Apc)])

    return data
          
```

```python
N = 10
nb_perm = 20
Co = []
for i in range(2,N-2):
    G = prime_factor(i)
    Co += k_random(nb_perm, G)
Co
```

```python
def mismatch(k, data):
    inc = []
    for _ in range(k):
        elt = random.choice(data)
        M = elt[0]
        A = random.choice(data)[1]
        if (A == elt[1]).all():
            continue
        else:
            inc.append([M, A])
    return inc

```

```python
Inc = mismatch(nb_perm, Co)
Inc
```

```python
# LABELED DATA
def labeled(Co, Inc):
    All_data = []
    for elt in Co:
        All_data.append(elt + [np.array(1)])
    for elt in Inc:
        All_data.append(elt + [np.array(0)])
    return All_data
```

```python
Data = labeled(Co, Inc) 
random.shuffle(Data)
len(Data)
```

```python
# Validation Dataset
Co_val = []
for i in range(N-2, N+1):
    G = prime_factor(i)
    Co_val += k_random(nb_perm, G)
    
Inc_val = mismatch(nb_perm, Co_val)

val_data = labeled(Co_val, Inc_val)
len(val_data)
```

```python
from sklearn.svm import SVC
from sklearn import metrics
from sklearn.model_selection import train_test_split
```

```python
X = [elt[0:-1] for elt in Data]
y = [elt[-1] for elt in Data]
X_test = [elt[0:-1] for elt in val_data]
y_test = [elt[-1] for elt in val_data]
X_train, X_test, y_train, y_test = np.array(X), np.array(X_test), np.array(y), np.array(y_test)

nsamples, nx, ny, nz = X_train.shape
d2_X_train = X_train.reshape((nsamples,nx*ny*nz))
nsamples, nx, ny, nz = X_test.shape
d2_X_test = X_test.reshape((nsamples,nx*ny*nz))
```

```python
model = SVC()
model.fit(d2_X_train, y_train)
```

```python
y_pred = model.predict(d2_X_test)

print("Precision:", metrics.accuracy_score(y_test, y_pred))
```
