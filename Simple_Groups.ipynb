```python
# Cayley table of all groups labeled 1 if simple and 0 if not
Cay_tab = []
for g in GG:
    C = gap.MultiplicationTable(g)
    C = np.array(C)
    if gap.IsSimpleGroup(g):
        Cay_tab.append([C, np.array([1])])
                       
    else:
        Cay_tab.append([C, np.array([0])])
len(Cay_tab)
        
```

```python
# fill all incomplet table
n = 30
for c in Cay_tab:
    if c[0].shape[0] < n:
        c[0] = fill(c[0], n)
Cay_tab
```

```python
# ROW/COL PERMUTATION FUNCTION
def row_col_permutation(mat):
    lin = list(mat)
    n = len(lin)
    idx = random.sample(list(range(n)), n)
    mat1 = [lin[i] for i in idx]
    col = list(zip(*mat1))
    mat2 = [col[i] for i in idx]
    return np.array(mat2).T
```

```python
k = 10^2
All_data = []
for c in Cay_tab:
    All_data.append(c)
    for i in range(k):
        P = row_col_permutation(c[0])
        All_data.append([P, c[1]])
        
len(All_data)        
```

```python
random.shuffle(All_data)
```

```python
X = []
y = []
for elt in All_data:
    X.append(elt[0])
    y.append(elt[1])



X_train, X_test, y_train, y_test = train_test_split(np.array(X),np.array(y), test_size=0.4, random_state=1)

nsamples, nx, ny = X_train.shape
d2_X_train = X_train.reshape((nsamples,nx*ny))
nsamples, nx, ny = X_test.shape
d2_X_test = X_test.reshape((nsamples,nx*ny))
```

```python
model = SVC()
model.fit(d2_X_train, y_train)
```

```python
y_pred = model.predict(d2_X_test)

print("Precision:", metrics.accuracy_score(y_test, y_pred))
```
