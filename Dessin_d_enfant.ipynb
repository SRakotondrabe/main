```python
def rotation(cycle):
    k = random.randint(0, len(cycle)-1)
    return cycle[k:]+ cycle[:k]
```

```python
def cycle(DATA, epoch):
    data = []
    for _ in range(epoch):
        for lst in DATA:
        #print(lst[0])
            new_cycle = []
            for cycle in lst[0]:
            
                if random.random()< 0.6:
                    new_cycle.append(rotation(cycle))
                else:
                    new_cycle.append(cycle)
        
            data.append([new_cycle, lst[1]]) 

    return DATA + data

```

```python
# Neural network
####################################################
nb_entree = 48
nb_neurone_caches = 20
nb_classe = 4
eta = 0.05
nb_epoch = 1000

np.random.seed(42)

# weights and bias

W1 = np.random.randn(nb_entree, nb_neurone_caches) * 0.1
b1 = np.zeros((1, nb_neurone_caches))

W2 = np.random.randn(nb_neurone_caches, nb_neurone_caches) * 0.1
b2 = np.zeros((1, nb_neurone_caches))

W3 = np.random.randn(nb_neurone_caches, nb_neurone_caches) * 0.1
b3 = np.zeros((1, nb_neurone_caches))

W4 = np.random.randn(nb_neurone_caches, nb_classe) * 0.1
b4 = np.zeros((1, nb_classe))
```

```python
# Activation Functions

def relu(Z):
    return np.maximum(0, Z)


def relu_derivee(Z):
    return (Z > 0).astype(float)


def softmax(Z):
    Z_stable = Z - np.max(Z, axis=1, keepdims=True) 
    exp_Z = np.exp(Z_stable)
    return exp_Z / np.sum(exp_Z, axis=1, keepdims=True)



```

```python
#  forward propagation

def propagation_avant(X):
    Z1 = X @ W1 + b1
    A1 = relu(Z1)

    Z2 = A1 @ W2 + b2
    A2 = relu(Z2)

    Z3 = A2 @ W3 + b3
    A3 = relu(Z3)

    Z4 = A3 @ W4 + b4
    A4 = softmax(Z4) 

    cache = (X, Z1, A1, Z2, A2, Z3, A3, Z4, A4)
    return A4, cache

```

```python
# Conversion des labels 1-4 
# 1 ---> [1,0,0,0]
# 2 ---> [0,1,0,0]
# 3 ---> [0,0,1,0]
# 4 ---> [0,0,0,1]

def convertir_en_one_hot(y):
    y = np.array(y).astype(int) - 1
    m = len(y)
    one_hot = np.zeros((m, nb_classe))
    one_hot[np.arange(m), y] = 1
    return one_hot

```

```python
# Calcul the loss function (Cross-Entropy)

def calculer_perte(A4, y_one_hot):
    m = y_one_hot.shape[0]
    epsilon = 1e-12 
    return -np.sum(y_one_hot * np.log(A4 + epsilon)) / m

```

```python
# Calcul gradient

def retropropagation(cache, y_one_hot):
    X, Z1, A1, Z2, A2, Z3, A3, Z4, A4 = cache
    m = y_one_hot.shape[0]

    # layer 4
    dZ4 = A4 - y_one_hot
    dW4 = A3.T @ dZ4 / m
    db4 = np.sum(dZ4, axis=0, keepdims=True) / m

    # layer 3
    dA3 = dZ4 @ W4.T
    dZ3 = dA3 * relu_derivee(Z3)
    dW3 = A2.T @ dZ3 / m
    db3 = np.sum(dZ3, axis=0, keepdims=True) / m

    # layer 2
    dA2 = dZ3 @ W3.T
    dZ2 = dA2 * relu_derivee(Z2)
    dW2 = A1.T @ dZ2 / m
    db2 = np.sum(dZ2, axis=0, keepdims=True) / m

    # layer 1
    dA1 = dZ2 @ W2.T
    dZ1 = dA1 * relu_derivee(Z1)
    dW1 = X.T @ dZ1 / m
    db1 = np.sum(dZ1, axis=0, keepdims=True) / m

    return dW1, db1, dW2, db2, dW3, db3, dW4, db4

```

```python
# Weights update

def mettre_a_jour_poids(gradients):
    global W1, b1, W2, b2, W3, b3, W4, b4
    dW1, db1, dW2, db2, dW3, db3, dW4, db4 = gradients

    W1 -= eta * dW1
    b1 -= eta * db1
    W2 -= eta * dW2
    b2 -= eta * db2
    W3 -= eta * dW3
    b3 -= eta * db3
    W4 -= eta * dW4
    b4 -= eta * db4

```

```python
def predire(X):
    A4, _ = propagation_avant(X)
    return np.argmax(A4, axis=1) + 1


def calculer_precision(X, y):
    predictions = predire(X)
    return np.mean(predictions == np.array(y))


```

```python
def entrainer(X, y, epochs=nb_epoch):
    y_one_hot = convertir_en_one_hot(y)

    for epoch in range(1, epochs + 1):
        A4, cache = propagation_avant(X)
        perte = calculer_perte(A4, y_one_hot)
        gradients = retropropagation(cache, y_one_hot)
        mettre_a_jour_poids(gradients)

        if epoch % (epochs // 10) == 0 or epoch == 1:
            precision = calculer_precision(X, y)
            print(f"Époque {epoch:4d}/{epochs} - perte : {perte:.4f} - précision : {precision:.2%}")

```
