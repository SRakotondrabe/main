```python
# ROW/COL permutation
def row_col_permutation(mat):
    lin = list(mat)
    n = len(lin)
    idx = random.sample(list(range(n)), n)
    mat1 = [lin[i] for i in idx]
    col = list(zip(*mat1))
    mat2 = [col[i] for i in idx]
    return np.array(mat2).T
```

```python
# GENERATE RANDOM LATIN SQUARE

def latin_square(n):
    if n <= 0:
        return []
    else:
        symbols = list(range(1,n+1))
        square = _latin_square(symbols)
        return _shuffle(square)

def _shuffle(matrix):
    square = deepcopy(matrix)
    random.shuffle(square)
    trans = list(zip(*square))
    random.shuffle(trans)
    return trans


def _latin_square(symbols):
    n = len(symbols)
    if n == 1:
        return [symbols]
    else:
        sym = random.choice(symbols)
        symbols.remove(sym)
        square = _latin_square(symbols)
        square.append(square[0].copy())
        for i in range(n):
            square[i].insert(i, sym)
        return square
```

```python
from sklearn.svm import SVC
from sklearn import metrics
from sklearn.model_selection import train_test_split
X = []
y = []
for elt in data:
    X.append(elt[0])
    y.append(elt[1])



X_train, X_test, y_train, y_test = train_test_split(np.array(X),np.array(y), test_size=0.4, random_state=1)

nsamples, nx, ny = X_train.shape
d2_X_train = X_train.reshape((nsamples,nx*ny))
nsamples, nx, ny = X_test.shape
d2_X_test = X_test.reshape((nsamples,nx*ny))
```

```python
model = SVC()
model.fit(d2_X_train, y_train)
```

```python
y_pred = model.predict(d2_X_test)

print("Precision:", metrics.accuracy_score(y_test, y_pred))
print("Matthews:", metrics.matthews_corrcoef(y_test, y_pred))
```
