```python
import numpy as np
import random
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn import metrics
```

```python
GG = gap.AllGroups(12) # All groups of size 12 
S1 = [] 
S2 = []
count = 0
for g in GG:
    if count < 3:
        S1.append(g)
        count+= 1
    else:
        S2.append(g)
```

```python
# ROW/COL PERMUTATION FUNCTION
def row_col_permutation(mat):
    lin = list(mat)
    n = len(lin)
    idx = random.sample(list(range(n)), n)
    mat1 = [lin[i] for i in idx]
    col = list(zip(*mat1))
    mat2 = [col[i] for i in idx]
    return np.array(mat2).T + 1
```

```python
T1 = []
nb_perm = 10
for g in S1:
    C = gap.MultiplicationTable(g)
    T1.append(np.array(C))

Table = []    
    
for c in T1:
    cls = [c]
    for i in range(nb_perm):
        Pr = row_col_permutation(c)
        cls.append(Pr)
    Table.append(cls)
        
```

```python
def _is_same_table(elt1, elt2, table):
    for cls in table:
        if (elt1 in cls) and (elt2 in cls):
            return 1
        else:
            return 0
```

```python
def match_data(Table, size):
    Data = []
    for _ in range(int(size/2)):
        cls = random.choice(Table)
        cls1 = random.choices(cls, k=2)
        Data.append(cls1 + [np.array(1)])
    for _ in range(int(size/2)):
        two_cls = random.choices(Table, k=2)
        elt1 = random.choice(two_cls[0])
        elt2 = random.choice(two_cls[1])
        Data.append([elt1, elt2, np.array(0)])
        
    return Data
        
            
    
    
    
    
```

```python
train_data = match_data(Table, size = 30)
train_data
```

```python
T2 = []
for g in S2:
    C = gap.MultiplicationTable(g)
    T2.append(np.array(C))

Table2 = []    
    
for c in T2:
    cls = [c]
    for i in range(nb_perm):
        Pr = row_col_permutation(c)
        cls.append(Pr)
    Table2.append(cls)
    
```

```python
val_data = match_data(Table2, size = 20)
```

```python
X_train = [elt[0:-1] for elt in train_data]
X_test = [elt[0:-1] for elt in val_data]
y_train = [elt[-1] for elt in train_data]
y_test = [elt[-1] for elt in val_data]


X_train, y_train, X_test, y_test = np.array(X_train), np.array(y_train), np.array(X_test), np.array(y_test)

nsamples, nx, ny, nz = X_train.shape
d2_X_train = X_train.reshape((nsamples,nx*ny*nz))
nsamples, nx, ny, nz = X_test.shape
d2_X_test = X_test.reshape((nsamples,nx*ny*nz))
```

```python
model = SVC()
model.fit(d2_X_train, y_train)
```

```python
y_pred = model.predict(d2_X_test)

print("Precision:", metrics.accuracy_score(y_test, y_pred))
print("Matthews:", metrics.matthews_corrcoef(y_test, y_pred))
```
